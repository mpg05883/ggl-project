# Results: Linear Regression Example

This notebook plots the training loss at each epoch from `losses.csv` and compares the learned weight and bias with the
true ones from `results.json`.

## Set Up

```python
import json

import matplotlib.pyplot as plt
import pandas as pd

from ggl.utils.path import resolve_results_dir, resolve_root_dir

EXPERIMENT = "2026-09-23_example"
RESULTS_DIR = resolve_results_dir(EXPERIMENT)
FIGURES_DIR = resolve_root_dir() / "experiments" / EXPERIMENT / "figures"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

## Training Loss

Plot the mean training loss at the end of each epoch.

```python
losses = pd.read_csv(RESULTS_DIR / "losses.csv")

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(losses["epoch"], losses["loss"])
ax.set_xlabel("Epoch")
ax.set_ylabel("Loss (MSE)")
ax.set_yscale("log")
ax.set_title("Training loss per epoch")
fig.savefig(FIGURES_DIR / "losses.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Learned vs. True Parameters

Show the learned weight and bias, the true weight and bias, and the final loss in one row.

```python
results = json.loads((RESULTS_DIR / "results.json").read_text())

columns = [
    "learned_weight",
    "true_weight",
    "learned_bias",
    "true_bias",
    "final_loss",
]
pd.DataFrame([{column: results[column] for column in columns}])
```
